# Problem 2: TD(0) State-Value Prediction on the Random Walk

7-state random walk **A–B–C–D–E–F–G**, start at **D**. A and G terminal.
Reward **+1** on transition into G, 0 otherwise. Behavior policy: left/right with prob 1/2. γ = 1.

True values for non-terminal states B..F:  **(1/6, 2/6, 3/6, 4/6, 5/6)**.

Parts (c) and (d): implement TD(0) and constant-α MC prediction, measure RMS error, and compare
learning curves across step sizes over many runs.

```
import numpy as np
import matplotlib.pyplot as plt

# State encoding: 0=A(term), 1=B, 2=C, 3=D(start), 4=E, 5=F, 6=G(term)
START = 3
TRUE = np.array([1/6, 2/6, 3/6, 4/6, 5/6])   # values for B..F
np.random.seed(0)
```

## (c) TD(0) and constant-α MC implementations

```
def run_episode():
    """One random walk. Returns list of (state, reward, next_state)."""
    s = START
    traj = []
    while True:
        s_next = s + np.random.choice([-1, 1])   # equiprobable left/right
        r = 1.0 if s_next == 6 else 0.0          # +1 only into G
        traj.append((s, r, s_next))
        s = s_next
        if s == 0 or s == 6:                     # terminal
            break
    return traj

def rms_error(V):
    """RMS error over the 5 non-terminal states B..F."""
    return np.sqrt(np.mean((V[1:6] - TRUE) ** 2))

def td0_prediction(n_episodes, alpha):
    V = np.full(7, 0.5); V[0] = 0.0; V[6] = 0.0    # terminals = 0
    errors = []
    for _ in range(n_episodes):
        for (s, r, s_next) in run_episode():
            V[s] += alpha * (r + V[s_next] - V[s])   # gamma = 1, TD(0) update
        errors.append(rms_error(V))
    return V, errors

def mc_prediction(n_episodes, alpha):
    V = np.full(7, 0.5); V[0] = 0.0; V[6] = 0.0
    errors = []
    for _ in range(n_episodes):
        traj = run_episode()
        G = sum(r for (_, r, _) in traj)   # gamma = 1: return from every visited state = total reward
        for (s, r, s_next) in traj:
            V[s] += alpha * (G - V[s])      # constant-alpha MC update
        errors.append(rms_error(V))
    return V, errors

# quick single-run sanity check (100 episodes)
V_td, e_td = td0_prediction(100, 0.1)
V_mc, e_mc = mc_prediction(100, 0.02)
print("TD(0)  V(B..F):", np.round(V_td[1:6], 3), " RMS:", round(e_td[-1], 4))
print("MC     V(B..F):", np.round(V_mc[1:6], 3), " RMS:", round(e_mc[-1], 4))
print("true   V(B..F):", np.round(TRUE, 3))
```

## (d) RMS error vs. episodes, averaged over ≥ 50 runs, multiple step sizes

```
def averaged_curve(method, alpha, n_episodes=100, n_runs=100):
    """Average the per-episode RMS-error curve over independent runs."""
    all_errs = np.zeros((n_runs, n_episodes))
    for run in range(n_runs):
        _, errs = method(n_episodes, alpha)
        all_errs[run] = errs
    return all_errs.mean(axis=0)

N_EPISODES = 100
N_RUNS = 100                      # >= 50 as required
TD_ALPHAS = [0.05, 0.10, 0.15]
MC_ALPHAS = [0.01, 0.02, 0.04]

np.random.seed(42)
td_curves = {a: averaged_curve(td0_prediction, a, N_EPISODES, N_RUNS) for a in TD_ALPHAS}
mc_curves = {a: averaged_curve(mc_prediction,  a, N_EPISODES, N_RUNS) for a in MC_ALPHAS}

plt.figure(figsize=(10, 6))
for a in TD_ALPHAS:
    plt.plot(range(1, N_EPISODES + 1), td_curves[a],
             linestyle="-", label=f"TD(0), α = {a}")
for a in MC_ALPHAS:
    plt.plot(range(1, N_EPISODES + 1), mc_curves[a],
             linestyle="--", label=f"MC, α = {a}")
plt.xlabel("Episodes")
plt.ylabel("RMS error (avg over B..F)")
plt.title(f"TD(0) vs constant-α MC on the random walk (avg of {N_RUNS} runs)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("td_vs_mc_rms.png", dpi=120, bbox_inches="tight")
plt.show()
```

### Discussion of (d)

**TD(0) beats MC here.** For every step size shown, the TD curves (solid) fall faster and settle at
a lower RMS error than the MC curves (dashed). This reproduces the classic Sutton & Barto Example 6.2
result. The reason is the **bias–variance trade-off**: TD's one-step target
$R_{t+1}+\gamma V(S_{t+1})$ depends on a single random transition, so it has low variance; MC's target
is the full return, whose variance compounds over the whole episode. On this short, well-behaved
Markov chain the small bias TD pays is far outweighed by the variance it saves, so it learns faster
and more stably.

**Effect of the step size α.**
- *Larger α* → faster initial drop but a **higher error floor**, because big updates keep the estimate
  jittering around the true values instead of settling (the estimate never stops chasing the latest
  noisy target). This is visible as the large-α curves crossing below the small-α ones early but
  plateauing higher.
- *Smaller α* → slower initial learning but a **lower, smoother asymptote**, since each update is a
  gentler correction and the running average is more stable.
- So α controls a speed-vs-final-accuracy trade-off. Neither method converges to zero error with a
  *constant* α (constant step sizes track a moving average and retain residual variance); reaching
  zero would require α to decay over time.

**Why the α ranges differ between methods.** MC needs much smaller step sizes to look stable because
its target has higher variance — a given α produces noisier updates for MC than for TD, so the MC
curves are compared at α ∈ {0.01, 0.02, 0.04} while TD uses {0.05, 0.10, 0.15}. Even so, TD at its
larger step sizes remains competitive with or better than MC at its smaller ones, underscoring TD's
lower per-update variance.
